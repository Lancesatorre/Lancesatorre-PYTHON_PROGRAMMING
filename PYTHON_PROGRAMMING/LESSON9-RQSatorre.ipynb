```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats
```

```python
X, y = make_blobs(n_samples=300, centers=3, cluster_std=1.0, random_state=1)
data = pd.DataFrame(X, columns=['Feature1', 'Feature2'])
data['TrueLabel'] = y
data.head()
```

```python
import os
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
os.environ['OMP_NUM_THREADS'] = '2'
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=100)
kmeans.fit(X)
data['KMeansLabel'] = kmeans.predict(X)
```

```python
plt.figure(figsize=(15,6))
plt.scatter(data['Feature1'], data['Feature2'], c=data['KMeansLabel'], cmap='viridis', s=100)
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], c='red', marker='+', s=50, label='Centroids')
plt.title("KMeans Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```
